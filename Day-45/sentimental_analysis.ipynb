```python

from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
import torch.nn.functional as F

# Load tokenizer and model (distilbert-base-uncased-finetuned-sst-2-english is a sentiment analysis model)
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Sample texts for classification
texts = [
    "I love learning NLP with Hugging Face!",
    "This is the worst movie I have ever seen.",
    "Transformers models are so powerful and easy to use.",
    "I'm not sure if I like this new phone.",
]

# Tokenize and encode inputs (batch)
encoded_inputs = tokenizer(texts, padding=True, truncation=True, return_tensors="pt")

# Forward pass
with torch.no_grad():
    outputs = model(**encoded_inputs)

# Get logits and apply softmax for probabilities
logits = outputs.logits
probs = F.softmax(logits, dim=1)

# Labels for SST-2 dataset: 0 = negative, 1 = positive
labels = ["Negative", "Positive"]

# Print results
for text, prob in zip(texts, probs):
    pred_label = labels[prob.argmax()]
    confidence = prob.max().item()
    print(f"Text: {text}")
    print(f"Predicted sentiment: {pred_label} (confidence: {confidence:.2f})")
    print("-" * 50)
```
